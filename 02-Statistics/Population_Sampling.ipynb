### Sampling a Data from the whole Population


### Import Library

```python
import pandas as pd #useful for loading the dataset
import numpy as np
```

### Load Dataset from Local Directory

```python
dataset = pd.read_csv(r'E:\Data Analytics Intern\Data_Analytics1\Day_8\Dataset.csv')
print(dataset.shape)
print(dataset.describe())
print(dataset.head(5))  
```

### Simple Random Sampling

```python
# take a simple random sample of 5 rows and sort by Id
simpleRandomSample = dataset.sample(n=5).sort_values(by='Id')

# calculate the mean SalePrice of the sampled rows
mean_simpleRandomSample = round(simpleRandomSample['SalePrice'].mean(), 3)

# display the mean of the simple random sample
print("Mean of SimpleRandomSample:", mean_simpleRandomSample)

# display the sampled rows
print(simpleRandomSample)
```

### Systematic Sampling

```python
# select every 3rd row from the dataset for systematic sampling
index = np.arange(0, len(dataset), step=3)
# print(index)

# create the systematic sample using the selected row indices
systematicSample = dataset.iloc[index]

# calculate the mean SalePrice for the systematic sample
mean_systematicSample = round(systematicSample['SalePrice'].mean(), 3)

# display the mean and the sampled rows
print("Mean of SystematicSample:", mean_systematicSample)
print(systematicSample)
```

### Cluster Sampling

```python
n = 5
# Divide the units into clusters of equal size
dataset['cluster_id'] = np.repeat(np.arange(1, n + 1), len(dataset) // n)

index = []
# Select rows from even-numbered clusters only
for i in range(len(dataset)):
    if dataset['cluster_id'].iloc[i] % 2 == 0:
        index.append(i)

clusterSample = dataset.iloc[index]
mean_clusterSample = round(clusterSample['SalePrice'].mean(), 3)

print("Mean of clusterSample:", mean_clusterSample)
print(clusterSample)
```

### Stratifield Random Sampling - Adding Subgroup(Strata)

```python
dataset = pd.read_csv(r'E:\Data Analytics Intern\Data_Analytics1\Day_8\Dataset.csv')  # loa
dataset['strata'] = np.repeat([1, 2], len(dataset) // 2)  # create two strata by repeating 1 and 2
index = []  # initialize an empty list to store selected row indices
for i in range(len(dataset)):
  index.append(i)  # add each row index to the list
stratifieldRandomSample = dataset.iloc[index]  # select rows using the collected indices
stratifieldRandomSample
```

### Stratifield Random Sampling - Sampling

```python
#!pip intall scikit-learn
from sklearn.model_selection import StratifiedShuffleSplit as sss

# Set the split criteria: one split and a test sample size of 8 rows
split = sss(n_splits=1, test_size=8)

# Perform stratified split using the 'strata' column
for x, y in split.split(dataset, dataset['strata']):
    stratifiedRandomSample = dataset.iloc[y].sort_values(by='SalePrice')

# Display the stratified random sample sorted by SalePrice
stratifiedRandomSample
```
