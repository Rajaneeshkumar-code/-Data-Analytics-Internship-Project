# 7_Probability Permutation & Combination


### Permutation - with Length-2

```python
from itertools import permutations

count = 0
data = [1, 2, 3]

# generate all permutations of length 2 from the list
perm = permutations(data, 2)

for i in list(perm):
    print(i)
    count += 1

print("Permutation Count: ", count)
```

### Combination - with Length-2

```python
from itertools import combinations  # import combinations from itertools
count = 0  # initialize counter
data = [1, 2, 3]
data2 = [2, 1, 3]  # list used for generating combinations
perm = combinations(data2, 2)  # generate all 2-length combinations from data2
for i in list(perm):
    print(i)  # print each combination
    count += 1  # increment the combination counter

print("Combination Count: ", count)  # print total count
```

### Combination - with Length-2

```python
from itertools import combinations_with_replacement as cr
count=0
data = [1, 2, 3]
data2 = [2, 1, 3]
perm = cr(data, 2)
for i in list(perm):
    print (i)
    count+=1

print("Combination with Element Count: ",count)
```

- This notebook covers basic permutation and combination examples using `itertools`.
- The permutation section imports `permutations`, then generates and prints all 2-length permutations of `[1, 2, 3]` and prints the count.
- The combination section imports `combinations`, then generates and prints all 2-length combinations of `[2, 1, 3]` and prints the count.
- The final section imports `combinations_with_replacement`, generates and prints all 2-length combinations with replacement from `[1, 2, 3]`, and prints the count.
- Variables used include `count`, `data`, `data2`, `perm`, and `i` to track iteration results and totals.
